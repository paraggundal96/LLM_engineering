# Financial Chatbot

## 📌 Overview
This chatbot provides financial insights based on predefined queries.

## 🚀 How to Run
1. Install Jupyter Notebook and pandas library.
2. open `chatbot.ipynb` in Jupyter Notebook
3. Follow the on-screen prompts.

## 📉 Limitations
- Only supports predefined financial metrics.
- Data is hardcoded for Microsoft only.
- Doesn't support natural language input.


```python
import pandas as pd
```

```python
df = pd.DataFrame(pd.read_excel('Finance.xlsx'))
```

## ChatBot for Microsoft

```python
microsoft_df = df[df['Company'] == 'Microsoft'].sort_values(by = 'Year')
microsoft_df
```

```python
microsoft_df.columns = df.columns.str.strip()
```

```python
def assets(year):
    result = microsoft_df[microsoft_df['Year'] == year]
    if not result.empty:
        return f"Total Assets in {year}: ${result['Total Assets'].values[0]:,.2f}"
    return "Year not found."

def liabilities(year):
    result = microsoft_df[microsoft_df['Year'] == year]
    if not result.empty:
        return f"Total Liabilities in {year}: ${result['Total Liabilities'].values[0]:,.2f}"
    return "Year not found."

def trend(parameter):
    rolling_mean = microsoft_df[parameter].rolling(window=3, min_periods=1).mean()
    if rolling_mean.iloc[-1] > rolling_mean.iloc[-2]:
        return "Increasing"
    else:
        return "Decreasing"

def percent_growth(parameter):
    percentage_growth = ((microsoft_df[parameter].pct_change()) * 100).iloc[-1]
    return f"{parameter} percentage growth: {percentage_growth:.2f}%"
```

```python

def assets(year):
    result = microsoft_df[microsoft_df['Year'] == year]
    if not result.empty:
        return f"Total Assets in {year}: ${result['Total Assets'].values[0]:,.2f}"
    return "Year not found."


def liabilities(year):
    result = microsoft_df[microsoft_df['Year'] == year]
    if not result.empty:
        return f"Total Liabilities in {year}: ${result['Total Liabilities'].values[0]:,.2f}"
    return "Year not found."


def trend(parameter):
    parameter = parameter.strip()  
    if parameter not in microsoft_df.columns:
        return "Invalid parameter."
    
    rolling_mean = microsoft_df[parameter].rolling(window=3, min_periods=1).mean()
    return "Increasing" if rolling_mean.iloc[-1] > rolling_mean.iloc[-2] else "Decreasing"


def percent_growth(parameter):
    parameter = parameter.strip()  
    if parameter not in microsoft_df.columns:
        return "Invalid parameter."
    
    percentage_growth = ((microsoft_df[parameter].pct_change()) * 100).iloc[-1]
    return f"{parameter} percentage growth: {percentage_growth:.2f}%"


while True:
    print("\nChoose an option:")
    print("1. Get Total Assets")
    print("2. Get Total Liabilities")
    print("3. Check Trend (Increasing/Decreasing)")
    print("4. Check Percentage Growth")
    print("5. Exit")
    
    choice = input("Enter your choice (1-5): ")
    
    if choice == '1':
        year = int(input("Enter year (e.g., 2020): "))
        print(assets(year))
    
    elif choice == '2':
        year = int(input("Enter year (e.g., 2020): "))
        print(liabilities(year))
    
    elif choice == '3':
        param = input("Enter financial parameter (Total Assets, Total Liabilities, Cash Flow, Net Income): ")
        print(f"{param} Trend: {trend(param)}")
    
    elif choice == '4':
        param = input("Enter financial parameter (Total Assets, Total Liabilities, Cash Flow, Net Income): ")
        print(percent_growth(param))
    
    elif choice == '5':
        print("Exiting the program.")
        break
    
    else:
        print("Invalid choice. Please enter a number between 1 and 5.")

```
