```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame(pd.read_excel('Finance.xlsx'))
type(data)
```

```python
data
```

## Rolling Data

```python
rolling_data = pd.DataFrame()
rolling_data[['Rolling Revenue','Rolling Net Income', 'Rolling total Assets', 'Rolling Liabilities', 'Rolling Cash Flow']] = data.groupby(['Company','Year'])[['Total Revenue', 'Net Income', ' Total Assets','Total Liabilities', 'Cash Flow']].rolling(window = 3,min_periods =1).mean()
```

```python
rolling_data
```

## Cummulative Growth

```python
cummulative = pd.DataFrame()
cummulative[['Cummulative Revenue','Cummulative Assets']] = data.groupby(['Company','Year'])[['Total Revenue',' Total Assets']].expanding().sum()
cummulative
```

## Pivot Table

```python
data.pivot_table(values = ['Total Revenue','Net Income'], index = 'Year', columns = 'Company', aggfunc = 'sum')
```

## Percentage Change

```python

data[['Total Revenue %', 'Net Income %', 'Total Assets %', 'Total Liabilities %', 'Cash Flow%']] = data.groupby('Company')[['Total Revenue', 'Net Income', ' Total Assets', 'Total Liabilities', 'Cash Flow']].pct_change() * 100
data
```
